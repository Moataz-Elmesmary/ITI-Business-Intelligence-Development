## 1. Function with Global and Local Variables

```python
total = 0

def add_to_total(n):
    global total
    total = total + n
    local_total = total
    print("global total: " + str(total) + ", Local total: " + str(local_total))
    
add_to_total(5)
```

## 2. Function with Global and Local Variables - Bank Balance

```python
balance = 1000

def withdraw(amount):
    global balance
    withdrawn_amount = amount
    balance = balance - amount
    print("Withdraw: " + str(withdrawn_amount) + ", balance: " + str(balance))
    
withdraw(100)
withdraw(200)
withdraw(400)
print("balance after withdraw: " + str(balance))
```

## 3. Function with Global and Local Variables - Scope

```python
count = 100

def modify_global_var():
    count = 10  
    print("Value of function inside the function: = " + str(count))

modify_global_var()


#__Outside
print("Value of function inside the function: = " + str(count))
```

## 4. Using enumerate in a List

```python
def print_items_with_index(items):
    for index, i in enumerate(items):
        print(str(index)+ " " + i)

Ex = ["apple", "banana", "cherry"]
print_items_with_index(Ex)
```

## 5. Using enumerate with a Start Index

```python
def print_items_with_index(items):
    for index, i in enumerate(items, start=1):
        print(str(index)+ " " + i)
        
Ex = ["apple", "banana", "cherry"]
print_items_with_index(Ex)
```

## 6. Lambda Function for Squaring Numbers

```python
square = lambda x : x ** 2
nums = [1, 2, 3]

squared_nums = list(map(square, nums))
print(squared_nums)
```

## 7. Lambda Function for Filtering Even Numbers
Example:
[1, 2, 3, 4, 5, 6] -> [2, 4, 6]

```python
even_nums = list(filter(lambda x : x % 2 == 0, [1, 2, 3, 4, 5, 6]))
print(even_nums)
```

## 8. Lambda Function for Reversing a String

```python
reverse_string = lambda x : x[::-1]
string = "Moataz"

print(reverse_string(string))
```

## 9. Create a Simple Class in Python

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

st1 = Student("Ahmed", 25)
print("Name: " + st1.name + ", Age: " + str(st1.age))
```

## 10. Class with Methods

```python
class student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def get_details(self):
        return "Name: " + self.name + ", Age: " + str(self.age)


st2 = student("Mahmoud", 20)
print(st2.get_details())
```
