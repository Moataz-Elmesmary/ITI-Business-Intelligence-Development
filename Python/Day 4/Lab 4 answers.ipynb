# Q3:

```python
x = pd.Series([2, 4, 6, 8, 10])
y = pd.Series([1, 3, 5, 7, 9])


print("Addition = \n", x + y)
print("Multiplication = \n", x * y)
print("division = \n", x / y)
print("Subtraction = \n", x - y)
```

# Q4:

```python
s1 = pd.Series([2, 4, 6, 8, 10])
s2 = pd.Series([1, 3, 5, 7, 10])

print("Equal:\n", s1 == s2)
print("Greater than:\n", s1 > s2)
print("Less than:\n", s1 < s2)
```

# Q7:

```python
df = pd.DataFrame({'Letter': ['A', 'A', 'B', 'B'], 
                   'Number': [1, 2, 1, 2], 
                   'Value': [10, 20, 30, 40]}).set_index(['Letter', 'Number'])

withFilter = df[df['Value'] > 10] #وبرده في filter / Query / where بس اخترت الاسرع

print(withFilter)
```
