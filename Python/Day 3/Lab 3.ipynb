## 1. Single Inheritance Task

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):  
    def __init__(self, name, age, student_id):
        super().__init__(name, age)  
        self.student_id = student_id

student1 = Student("Ahmed", 25, "123")
print(student1.name) 
print(student1.student_id) 
```

## 2. Multiple Inheritance Task

```python
class Sports:
    def __init__(self, sports_score):
        self.sports_score = sports_score

class Academics:
    def __init__(self, academic_score):
        self.academic_score = academic_score

class Student(Sports, Academics):  
    def __init__(self, sports_score, academic_score):
        Sports.__init__(self, sports_score)
        Academics.__init__(self, academic_score)
    
    def total_score(self):
        return self.sports_score + self.academic_score


std = Student(90, 95)
print("the total achievements of the student in sports and academics: " ,std.total_score()) 
```

## 3. Multilevel Inheritance Task

```python
class Animal:
    def eat(self):
        return "Animals eat other animals"

class Mammal(Animal):
    def breed(self):
        return "Mammals produce young animals"

class Dog(Mammal):
    def bark(self):
        return "Dog barks"
        
dog1 = Dog()
print(dog1.eat()) 
print(dog1.breed())  
print(dog1.bark())         
```

## 5. Encapsulation Task

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0  
    
    def deposit(self, amount):
        self.__balance = self.__balance + amount
    
    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance = self.__balance - amount
        else:
            print("Insufficient balance.")

            
account = BankAccount()
account.deposit(1000)
account.withdraw(500)  
account.withdraw(700) 
```

### 1. Can We Do Overloading in Python?

No, not like other languages, but we can mimic it using these techniques:

(1) Default Parameters:

```python
class Calculator:
    def add(self, a, b, c=0): 
        return a + b + c

calc = Calculator()
print(calc.add(1, 2))     # c = 0
print(calc.add(1, 2, 3))  
```

(2) Variable-Length Arguments (*args/**kwargs)

```python
class Calculator:
    def add(self, *args):
        return sum(args)

calc = Calculator()
print(calc.add(1, 2))       
print(calc.add(1, 2, 3, 4)) 
```

(3) Type-Based Overloading with functools.singledispatch

```python
from functools import singledispatch

@singledispatch
def process(data):
    raise NotImplementedError("Unsupported type")

@process.register(int)
def _(data):
    return f"Processing integer: {data}"

@process.register(str)
def _(data):
    return f"Processing string: {data}"

print(process(10))   
print(process("Hi"))
```

### 2. protected access modifier and clarify your searching with example and what is different between protected and public in python

1. Protected Access Modifier

- Single underscore prefix (_).

- Accessibility: Meant for internal use within the class or its subclasses. Not enforced by Python.

Example:

```python
class Base:
    def __init__(self):
        self._protected_attr = 42 

class Derived(Base):
    def get_protected(self):
        return self._protected_attr  ### Accessible in subclass

obj = Derived()
print(obj.get_protected())

### Still accessible outside the class
print(obj._protected_attr)
```

2. Public Access Modifier
- No underscores in the name.

- Accessible everywhere: Inside the class, outside the class, and in subclasses.

Example:

```python
class Person:
    def __init__(self, name):
        self.name = name      # (Public attribute)

p = Person("Ahmed")
print(p.name)  
```
